# Phishing URL Detection Training Notebook
This notebook performs dataset analysis, feature extraction, model training, evaluation, and model saving for the phishing URL detection system.

```python
import sys
from pathlib import Path
BASE_DIR = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
if str(BASE_DIR) not in sys.path:
    sys.path.insert(0, str(BASE_DIR))

import pandas as pd
from feature_extraction import extract_features_from_dataframe
from training import load_dataset, preprocess_dataset, analyze_dataset, create_visualizations, train_random_forest

df = load_dataset('dataset/phishing.csv')
analyze_dataset(df)
```

```python
df_clean = preprocess_dataset(df)
features = extract_features_from_dataframe(df_clean)
display(features.head())
print('Feature columns count:', len(features.columns))
print('Feature columns:', features.columns.tolist())
```

```python
visualization_dir = 'static/images'
create_visualizations(df_clean, features, visualization_dir)
print('Visualizations saved to', visualization_dir)
```

```python
train_results = train_random_forest(df_clean, 'models/phishing_rf.joblib', 'static/images')
print('\nModel Training & Evaluation Results:')
for k, v in train_results.items():
    if k != 'confusion_matrix':
        print(f'  {k}: {v}')
print('  confusion_matrix:', train_results['confusion_matrix'])
```
